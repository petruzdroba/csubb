```python
#In cate moduri pot fi aranjati 5 oameni pe 12 scaune, tinand in minte ca trebuie sa fie minim un scaun intre ei

#Asezam 5 persoane pe 8 locuri, si dupa inseram scaune inainte sau dupa toate persoanele
pozitii = list(combinations(range(8), 5))
```

```python
perm_pers = list(permutations("12345"))
pers ="12345"
contor=0
toate_asezarile=[]
for pozitie in pozitii:
    scaune = list("_" * 8)
    for pers in perm_pers:
        for i in range(5):
            scaune[pozitie[i]] = pers[i]
        toate_asezarile.append(scaune)
        contor+=1
print(len(toate_asezarile))
toate_asezarile[3000]
```

```python
# insert
#parcurgem scaune si cand gasim ceva !="_"
scaune=toate_asezarile[3000]
scaune_noi = scaune
for i in range(len(scaune)):
    if scaune[i] != '_':
        scaune_noi.insert(i+1, '_')

print(scaune_noi)
```
