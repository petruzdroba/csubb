```python
from scipy.stats import norm
from numpy import mean, std, linspace
from matplotlib.pyplot import show, hist, grid, legend, xticks, plot
```

```python
m = 165 # inaltimea medie
s = 10 #deviatia standard - cat de mult deviaza celalalte inaltimi de medie

data = norm.rvs(loc=m, scale=s, size=1000)
```

```python
hist(data, bins=16, density = True, range=(130,210), rwidth = 0.9, color = 'Orchid', edgecolor = 'RebeccaPurple')
xticks(range(130,210,5), fontsize=8)
grid()

x = linspace(130,210, 1000)
plot(x, norm.pdf(x, loc=m, scale=s), color='RebeccaPurple')

show()
```

```python
print(mean(data)) # media aritmetica dintre toate elementele din data
print(std(data))

print(mean([ 160 < x < 170 for x in data])) # vector boolean -> nnumar de omani care au inaltimea in acelasi interval si ii faci media aritmetica P(160 < X < 170)
print( norm.cdf(170, loc=m, scale=s) ) #P(X<=170)
print( norm.cdf(170, loc=m, scale=s) - norm.cdf(160, loc=m, scale=s) ) 
```

```python
from scipy.stats import uniform, expon
```

```python
def timp_printare():
    y = uniform.rvs()

    if y<= 0.4:
        return expon.rvs(loc=0, scale=1/(1/5)) # exponentiala 0 si inversul distributiei
    else:
        return uniform.rvs(loc=4, scale=6-4) # la uniforma capatul din stanga 
data = [ timp_printare() for _ in range(0, 1000)]
# print(data)

print(mean(data)) 
print(std(data))

print( mean([ x <= 5 for x in data])) # probabilitatea estimata ca timpul sa fie mai mi decat 5
print(0.4 * expon.cdf(5, loc=0, scale=1/(1/5)) + 0.6 * uniform.cdf(5, loc=4, scale=6-4)) # P(T<5) = P(T<5|I1)*P(I1) + P(T<5|I2)*P(I2) 
```
